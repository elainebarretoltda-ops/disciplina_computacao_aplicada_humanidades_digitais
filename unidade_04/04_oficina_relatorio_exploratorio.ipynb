As escolhas de limpeza e tokenização, incluindo remoção de pontuação, conversão para minúsculas e exclusão de stopwords, podem alterar as frequências e associações lexicais. Da mesma forma, os agrupamentos temporais, as faixas de páginas e os critérios de identificação de valores extremos influenciam os padrões quantitativos observados.

Por se tratar de uma base institucional, os resultados refletem o recorte documental, as temáticas e as características acadêmicas do programa. Não foram utilizados dados fictícios. Assim, diferenças de frequência ou extensão podem decorrer da composição do corpus, sem representar mudanças conceituais.

Os resultados não permitem estabelecer relações causais, generalizar conclusões para outros programas ou afirmar tendências históricas consolidadas. Como a análise conceitual ainda está em desenvolvimento, também não é possível concluir definitivamente como o conceito de Humanidades Digitais foi construído nas dissertações.
