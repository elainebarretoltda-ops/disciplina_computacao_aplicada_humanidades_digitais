**Estimativa de volume e trabalho de coleta:** O corpus previsto é composto por 26 dissertações do PPGIHD/UFRRJ, identificadas até outubro de 2026 e disponibilizadas no RIMA, repositório institucional da UFRRJ. O trabalho compreenderá a coleta e organização dos documentos, o registro dos metadados bibliográficos, a identificação dos trechos relacionados ao conceito de Humanidades Digitais e a análise das definições e referências teóricas mobilizadas. A estimativa de tempo dependerá da extensão dos documentos, das condições de extração textual e dos procedimentos de análise adotados.

**Amostra piloto:** Será realizada uma análise piloto com um subconjunto das 26 dissertações do PPGIHD/UFRRJ, disponíveis no RIMA. Essa etapa terá como objetivo testar os procedimentos de identificação, extração e categorização das definições de Humanidades Digitais, verificando a adequação dos indicadores e das categorias analíticas. Os resultados permitirão identificar dificuldades, ajustar os critérios de classificação e aprimorar o protocolo antes da análise do corpus completo. A quantidade de dissertações e os critérios de seleção da amostra piloto ainda serão definidos.

**Dependências de autorização ou infraestrutura:** A pesquisa dependerá do acesso às 26 dissertações do PPGIHD/UFRRJ disponibilizadas no RIMA, da possibilidade de obtenção dos arquivos integrais e da disponibilidade de ferramentas para organização, extração e análise textual. Será necessário verificar as condições de reutilização dos documentos e eventuais exigências de autorização. A infraestrutura computacional e as ferramentas específicas de análise ainda serão definidas conforme as necessidades metodológicas da pesquisa.

**Plano B se a principal fonte não puder ser usada:** As 26 dissertações que compõem o corpus da pesquisa estão disponíveis publicamente no RIMA, repositório institucional da UFRRJ, não sendo necessária autorização para acessá-las. As principais dependências estão relacionadas à infraestrutura computacional para armazenamento, organização, extração e análise dos dados textuais. As ferramentas específicas serão definidas conforme as necessidades metodológicas da pesquisa, considerando sua capacidade de identificar, sistematizar e comparar as diferentes construções do conceito de Humanidades Digitais presentes nas dissertações.

**O que será deliberadamente deixado de fora:** A pesquisa não incluirá dissertações de outros programas de pós-graduação, teses, artigos científicos ou outras produções acadêmicas como parte do corpus principal. Essas publicações poderão ser utilizadas na fundamentação teórica. Também não serão realizadas avaliações da qualidade acadêmica das dissertações ou do desempenho individual de seus autores. A investigação permanecerá concentrada na construção do conceito de Humanidades Digitais nas 26 dissertações identificadas no PPGIHD/UFRRJ, sem pretensão de estabelecer uma definição universal do campo.
