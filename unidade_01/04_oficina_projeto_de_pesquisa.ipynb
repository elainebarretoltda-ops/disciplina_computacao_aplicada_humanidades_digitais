## U01-A13 — Atividade — revisão entre pares

**Modalidade:** duplas, com dois turnos. **Tempo sugerido:** 15 minutos por
proposta. **Produto:** parecer breve recebido e registro das mudanças
realizadas.

No primeiro turno, o autor apresenta a proposta por até 3 minutos; o
revisor lê a síntese, responde às cinco perguntas e oferece um comentário
prioritário. Depois, invertam os papéis. Não reescreva a proposta do
colega: indique o ponto que precisa de decisão ou justificativa.

Peça a uma colega ou a um colega que responda:

1. consigo identificar exatamente o que será observado?
2. os dados propostos podem responder à pergunta?
3. conceito e indicador estão diferenciados?
4. o alcance da conclusão está claro?
5. qual escolha precisa de maior justificativa?

Revise a formulação após receber o comentário. A pergunta resultante encerra
a Unidade 1 e orientará a construção da base na Unidade 2.

### Parecer recebido

A proposta apresenta uma pergunta de pesquisa clara e um recorte institucional bem definido, permitindo identificar que serão observadas as construções conceituais de Humanidades Digitais nas dissertações do PPGIHD/UFRRJ. Os dados propostos são pertinentes à pergunta, pois possibilitam examinar definições, referências teóricas e abordagens conceituais. O conceito central está diferenciado dos indicadores, que incluem definições explícitas e implícitas, autores citados e categorias conceituais. O alcance das conclusões está adequadamente delimitado ao programa investigado, sem pretensão de generalização para todo o campo das Humanidades Digitais. Entretanto, considero necessário justificar melhor os critérios utilizados para identificar e classificar as definições implícitas, pois essa decisão poderá influenciar diretamente os resultados da análise.

### Mudanças realizadas e justificativa

A partir das observações apresentadas na simulação, propõe-se aprimorar os critérios de identificação e categorização das definições de Humanidades Digitais, estabelecendo uma distinção entre definições explícitas, implícitas e usos contextuais do termo. Também será necessário documentar as regras de classificação e preservar os trechos originais para permitir a verificação das interpretações. Esses ajustes buscam reduzir a subjetividade da análise, ampliar a transparência metodológica e fortalecer a relação entre os dados coletados e a pergunta de pesquisa.

Pergunta final mantida: Como o conceito de Humanidades Digitais tem sido construído no Programa de Pós-Graduação Interdisciplinar em Humanidades Digitais da UFRRJ?

A pergunta permanece adequada, pois o principal aprimoramento necessário está nos procedimentos metodológicos, e não na delimitação do problema de pesquisa.
